```python
#1. list creation
age_list = [24, 25, 27, 28, 29]
print("age_list:", age_list)
name_list = ["rakshan","raghav","pranish","nikhilan","vihaan"]
print("name_list:",name_list)
#2. List Operations / Modifications
name_list.append("yazhili")
print("\nappend name list:",name_list)
age_list.insert(2,26)
print("insert age:",age_list)
name_list.remove("yazhili")
print("after_removing:",name_list)
age_list.pop()
print("after pop:",age_list)
age_list.extend([30,31,32])
print("extend age list:",age_list)
age_list.sort(reverse = True )
print("sort age list:",age_list)
max(age_list)
print('maximum age:',max(age_list))
min(age_list)
print("minimun age:",min(age_list))
sum(age_list)
print("sum age list:",sum(age_list))
```

```python
#3. Accessing List Elements
name_list
name_list[0]
print("first name:",name_list[0])
name_list[4]
print("last name:",name_list[4])
name_list[2:5]
print("index 2 to 4:",name_list[2:5])
name_list[::-1]
print("reverse name list:",name_list[::-1])
```

```python
#Dictionary (Creation, Modification and Access)
student_marks = {
    "rakshan":85,
    "raghav":76,
    "pranish":54,
    "nikhilan":67,
    "vihaan":86
}
print("rakshan's_marks =",student_marks["rakshan"])
student_marks["janani"]= 80
student_marks["pranish"] = 82
print("student's mark=",student_marks)
print("keys=",student_marks.keys())
print("values=",student_marks.values())
print("items=",student_marks.items())
```

```python
#Sets (Operations)
my_set ={ 'a','e','i','o','u','a','a','i'}
print("my_set:",my_set)
#print(my_set[4])
#Reason: Sets are unordered, so we cannot access their elements using an index like [4].
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}  
print("union:",set1.union(set2))
print("intersection:",set1.intersection(set2))
```
